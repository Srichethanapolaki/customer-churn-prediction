```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/Telco-Customer-Churn.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
import pandas as pd

# Churn rate by contract type
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print("Churn Rate by Contract Type (%)")
print(contract_churn.round(2))

# Churn rate by technical support
support_churn = pd.crosstab(
    df["TechSupport"],
    df["Churn"],
    normalize="index"
) * 100

print("\nChurn Rate by Technical Support (%)")
print(support_churn.round(2))

# Churn rate by tenure
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, float("inf")],
    labels=["0–12 months", "13–24 months",
            "25–48 months", "49+ months"]
)

tenure_churn = pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100

print("\nChurn Rate by Tenure (%)")
print(tenure_churn.round(2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Calculate churn statistics
total_customers = len(df)
churned_customers = (df["Churn"] == "Yes").sum()
retained_customers = (df["Churn"] == "No").sum()
churn_rate = (churned_customers / total_customers) * 100

print("CUSTOMER CHURN SUMMARY")
print("-" * 35)
print("Total customers:", total_customers)
print("Customers who churned:", churned_customers)
print("Customers retained:", retained_customers)
print(f"Overall churn rate: {churn_rate:.2f}%")
```

```python
# Churn rate by contract type
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

print("Churn rate by contract type:")
print(contract_churn.round(2))

# Churn rate by internet service
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

print("\nChurn rate by internet service:")
print(internet_churn.round(2))

# Churn rate by payment method
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

print("\nChurn rate by payment method:")
print(payment_churn.round(2))
```

```python
# Calculate churn rate by contract type
contract_rate = (
    df.assign(Churned=(df["Churn"] == "Yes"))
      .groupby("Contract")["Churned"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
sns.barplot(x=contract_rate.index, y=contract_rate.values)

plt.title("Customer Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.show()
```

## Business Insights and Recommendations

### 1. Contract Type Analysis
Month-to-month customers have the highest observed churn rate at 42.71%. The churn rate is 11.27% for one-year contracts and 2.83% for two-year contracts.

**Recommendation:** Investigate the reasons for churn among month-to-month customers and evaluate whether appropriate long-term contract incentives could improve retention.

### 2. Technical Support Analysis
Customers without technical support have an observed churn rate of 41.64%, compared with 15.17% for customers with technical support.

**Recommendation:** Investigate support-related customer experiences and consider proactive assistance for customers who may need additional help.

### 3. Customer Tenure Analysis
Customers with 0–12 months of tenure have the highest observed churn rate at 47.44%. The rate declines to 28.71% for customers with 13–24 months of tenure, 20.39% for customers with 25–48 months, and 9.51% for customers with at least 49 months of tenure.

**Recommendation:** Strengthen customer onboarding and monitor early customer experiences to identify potential churn risks.

### Overall Conclusion
The analysis identifies contract type, technical support status, and customer tenure as useful variables for investigating customer churn. The results suggest that early-tenure customers and month-to-month customers deserve particular attention in retention strategies.

These are descriptive associations, not proof that any individual factor causes churn. Further analysis and machine learning evaluation are needed to assess predictive usefulness.

```python
# Calculate churn rate by payment method

payment_churn = (
    df.assign(Churned=(df["Churn"] == "Yes").astype(int))
      .groupby("PaymentMethod")["Churned"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

print("Churn Rate by Payment Method (%):")
print(payment_churn.round(2))

plt.figure(figsize=(10, 5))

ax = sns.barplot(
    x=payment_churn.index,
    y=payment_churn.values
)

ax.set_title("Customer Churn Rate by Payment Method")
ax.set_xlabel("Payment Method")
ax.set_ylabel("Churn Rate (%)")
ax.tick_params(axis="x", rotation=20)

for i, value in enumerate(payment_churn.values):
    ax.text(i, value + 1, f"{value:.2f}%", ha="center")

plt.ylim(0, min(100, max(payment_churn.values) + 10))
plt.tight_layout()
plt.show()
```
